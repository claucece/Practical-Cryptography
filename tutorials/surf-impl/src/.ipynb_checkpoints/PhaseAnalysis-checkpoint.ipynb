# Surf 3PH Phase Analysis

Joins the **verifier** timing CSV with the **MultiGet** client CSVs for one benchmark location
(`lan/`, `london/`, `va/`, or `seoul/`), drops bad runs, and reports the average **online** and
**offline** time and bandwidth per site.

- `verifier.csv` holds per-run phase breakdowns (`*_ms`, `*_bytes`, `*_rounds`), keyed by `client_random`.
- `MultiGetBench.csv` (HTTP/1.1) and `MultiGetH2Bench.csv` (HTTP/2) hold the `domain` (site) and an
  `empty_response` flag, also keyed by `client_random`.
- The join attaches the `domain` (and the bad-run flag) to each verifier run.

**Online phases** (interactive, during the live connection):
`KS_WAIT`, `KS_DONE`, `CERT_WAIT`, `CERT_WRITE`, `DERIVE_TS`, `DERIVE_RES`, `DERIVE_PSK`.

**Offline phases**: every other measured phase (preprocessing / setup).

```python
import pandas as pd
from pathlib import Path

# ---- Pick the benchmark location: 'lan', 'london', 'va', or 'seoul' ----
DATA_DIR = Path('lan')

# Phase names classified as ONLINE (interactive, during the live connection).
ONLINE_PHASES = [
    'KS_WAIT', 'KS_DONE', 'CERT_WAIT', 'CERT_WRITE',
    'DERIVE_TS', 'DERIVE_RES', 'DERIVE_PSK',
]

# The two client-side MultiGet benchmark files and the protocol label each represents.
MULTIGET_FILES = {
    'MultiGetBench.csv':   'h1',
    'MultiGetH2Bench.csv': 'h2',
}

assert DATA_DIR.is_dir(), f'{DATA_DIR} not found (run from src/)'
```

## 1. Load & combine the MultiGet (client) CSVs

Union the HTTP/1.1 and HTTP/2 client logs, tagging each with a `protocol` column.

```python
mg_frames = []
for fname, proto in MULTIGET_FILES.items():
    fpath = DATA_DIR / fname
    if not fpath.exists():
        print(f'skip (missing): {fpath}')
        continue
    df = pd.read_csv(fpath)
    df['protocol'] = proto
    mg_frames.append(df)

multiget = pd.concat(mg_frames, ignore_index=True)

# Data hygiene: some logs contain blank trailing rows (null client_random) and, occasionally,
# exact-duplicate rows. Drop both so client_random is a unique key for the join.
raw = len(multiget)
multiget = (multiget.dropna(subset=['client_random'])
                    .drop_duplicates('client_random', keep='first')
                    .reset_index(drop=True))
print(f'MultiGet rows: {len(multiget)} (dropped {raw - len(multiget)} null/duplicate row(s))')
print('protocols:', multiget["protocol"].value_counts().to_dict())
print('sites:', sorted(multiget["domain"].unique()))
multiget.head()
```

## 2. Load the verifier CSV and join on `client_random`

Each 3PH run produces a unique `client_random` that appears once in the verifier log and once in a
MultiGet log. An **inner** join keeps only runs present in both (i.e. runs that produced a client
record), and attaches the `domain`, `protocol`, and `empty_response` flag to the phase timings.

```python
verifier = pd.read_csv(DATA_DIR / 'verifier.csv')
# The verifier log can contain exact-duplicate rows (the same run written twice); collapse them
# so client_random is unique on this side too.
raw = len(verifier)
verifier = (verifier.dropna(subset=['client_random'])
                    .drop_duplicates()
                    .drop_duplicates('client_random', keep='first')
                    .reset_index(drop=True))
print(f'verifier rows: {len(verifier)} (dropped {raw - len(verifier)} null/duplicate row(s))')

# Columns to pull over from the client side.
mg_cols = ['client_random', 'domain', 'protocol', 'empty_response',
           'response_bytes', 'verifier_connect_ms', 'preprocess_ms',
           'origin_connect_ms', 'total_connect_ms']
mg_cols = [c for c in mg_cols if c in multiget.columns]

# run_type exists on both sides; keep the verifier's and avoid a merge collision.
joined = verifier.merge(
    multiget[mg_cols],
    on='client_random',
    how='inner',
    validate='one_to_one',
)
print(f'joined rows: {len(joined)}')
joined[['client_random', 'run_type', 'domain', 'protocol', 'empty_response']].head()
```

## 3. Drop bad runs

A run is **bad** if the client got an empty/failed response (`empty_response != 0` or
`response_bytes == 0`), or if it is missing a key online-phase measurement. These are the
handshake-timeout / failed-verifier cases that must not pollute the averages.

```python
before = len(joined)

bad = pd.Series(False, index=joined.index)
if 'empty_response' in joined:
    bad |= joined['empty_response'].fillna(1).astype(int) != 0
if 'response_bytes' in joined:
    bad |= joined['response_bytes'].fillna(0) <= 0

# A completed run should have spent time in the online phases; all-zero => it died early.
online_ms_cols = [f'{p}_ms' for p in ONLINE_PHASES]
bad |= joined[online_ms_cols].fillna(0).sum(axis=1) <= 0

clean = joined[~bad].copy()
print(f'dropped {before - len(clean)} bad run(s); {len(clean)} clean run(s) remain')
if (before - len(clean)) > 0:
    display(joined[bad][['client_random', 'domain', 'protocol', 'run_type', 'empty_response', 'response_bytes']])
```

## 4. Compute per-run online / offline time and bandwidth

`online_*` = sum over the online phases; `offline_*` = sum over every **other** measured phase.
Phase base names are discovered from the `*_ms` columns so the split always covers all phases.

```python
# Discover verifier phase base names: a real phase has BOTH a *_ms and a *_bytes column.
# This deliberately excludes the client-side timings merged in from the MultiGet CSV
# (preprocess_ms, origin_connect_ms, total_connect_ms, verifier_connect_ms), which have no
# *_bytes counterpart and would otherwise double-count into the offline total.
all_phases = [c[:-3] for c in clean.columns
              if c.endswith('_ms') and c != 'total_ms' and f'{c[:-3]}_bytes' in clean.columns]
offline_phases = [p for p in all_phases if p not in ONLINE_PHASES]

def phase_sum(df, phases, suffix):
    cols = [f'{p}_{suffix}' for p in phases if f'{p}_{suffix}' in df.columns]
    return df[cols].fillna(0).sum(axis=1)

clean['online_ms']     = phase_sum(clean, ONLINE_PHASES,  'ms')
clean['offline_ms']    = phase_sum(clean, offline_phases, 'ms')
clean['online_bytes']  = phase_sum(clean, ONLINE_PHASES,  'bytes')
clean['offline_bytes'] = phase_sum(clean, offline_phases, 'bytes')

print('ONLINE phases :', ONLINE_PHASES)
print('OFFLINE phases:', offline_phases)
clean[['domain', 'protocol', 'run_type',
       'online_ms', 'offline_ms', 'online_bytes', 'offline_bytes']].head()
```

## 5. The single large joined table

Every clean run, with its site, protocol, run type, full phase breakdown, and the derived
online/offline totals.

```python
lead = ['experiment_time', 'domain', 'protocol', 'run_type', 'client_random',
        'online_ms', 'offline_ms', 'online_bytes', 'offline_bytes']
lead = [c for c in lead if c in clean.columns]
big_table = clean[lead + [c for c in clean.columns if c not in lead]]
print(f'big_table: {big_table.shape[0]} rows x {big_table.shape[1]} cols')
big_table.head()
```

## 6. Per-site averages

Average online/offline **time (ms)** and **bandwidth (bytes)** for each site. Broken out by
`run_type` (full vs. resumed) since resumed runs skip work; set `GROUP_COLS` to just `['domain']`
to collapse across run types.

```python
GROUP_COLS = ['domain', 'run_type']  # e.g. ['domain'] to average across run types

metrics = ['online_ms', 'offline_ms', 'online_bytes', 'offline_bytes']
summary = (clean.groupby(GROUP_COLS)[metrics]
                .mean()
                .round(1))
summary['n_runs'] = clean.groupby(GROUP_COLS).size()

# Convenience unit conversions.
summary['online_s']    = (summary['online_ms']  / 1000).round(3)
summary['offline_s']   = (summary['offline_ms'] / 1000).round(3)
summary['online_MB']   = (summary['online_bytes']  / 1e6).round(2)
summary['offline_MB']  = (summary['offline_bytes'] / 1e6).round(2)

summary = summary[['n_runs', 'online_s', 'offline_s', 'online_MB', 'offline_MB',
                   'online_ms', 'offline_ms', 'online_bytes', 'offline_bytes']]
summary
```

```python
# Optional: write the joined table and the per-site summary out to CSV.
big_table.to_csv(DATA_DIR / 'joined_runs.csv', index=False)
summary.to_csv(DATA_DIR / 'per_site_summary.csv')
print(f'wrote {DATA_DIR / "joined_runs.csv"} and {DATA_DIR / "per_site_summary.csv"}')
```
